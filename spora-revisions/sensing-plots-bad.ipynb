```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
temp_data = pd.read_excel('published_v1 no-sense-num-glia-redo-spreadsheet.xlsx')
temp_data.head()
```

```python
temp_data = np.transpose(temp_data)
temp_data.head()
```

```python
temp_data = temp_data.filter([10,15])
temp_data.drop(['BehaviorSpace results (NetLogo 6.4.0)'], inplace=True)
temp_data.rename(columns={10:'Init-Microglia'}, inplace=True)
temp_data.rename(columns={15:'Ticks'}, inplace=True)
temp_data = temp_data.reset_index()
temp_data = temp_data.drop(['index'], axis=1)
temp_data.head()
```

```python
print(temp_data['Ticks'].min())
print(temp_data['Ticks'].max())
```

```python
plt.scatter(temp_data['Init-Microglia'],
            temp_data['Ticks'],
            alpha=0.4)

plt.style.use('bmh')
plt.ylabel('hours')
plt.xlabel('number of microglia')
plt.ylim(0, 9900)
plt.xlim(0, 21)
plt.xticks(np.arange(0, 21, 5))

plt.savefig('num-glia-no-sens-hrs.png', bbox_inches='tight')
```
