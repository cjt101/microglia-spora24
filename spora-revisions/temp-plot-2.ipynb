```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
temp_data = pd.read_excel('republished_v1 temp-redo-spreadsheet-2.xlsx')
temp_data.head(20)
```

```python
temp_data = np.transpose(temp_data)
temp_data.head(20)
```

```python
temp_data[13].value_counts()
```

```python
temp_data = temp_data.filter([13, 15])
temp_data.drop(['BehaviorSpace results (NetLogo 6.4.0)'], inplace=True)
temp_data.rename(columns={13:'Temperature'}, inplace=True)
temp_data.rename(columns={15:'Ticks'}, inplace=True)
temp_data.head()
```

```python
temp_data = temp_data.reset_index()
temp_data.head()
```

```python
temp_data = temp_data.drop(['index'], axis=1)
temp_data.head()
```

```python
temp_data = temp_data.groupby('Temperature').mean()
temp_data.head()
```

```python
plt.style.use('bmh')
plt.rcParams['figure.dpi'] = 300

(
    temp_data
    .plot()
)

plt.legend(['Average over 50 trials'])
plt.xlabel('temperature (C\xb0)')
plt.ylabel('hours')
plt.title('Graph 2')
plt.ylim(700, 1400)
plt.xlim(35.9, 40.1)

plt.savefig('p2_repubbed.png', bbox_inches='tight')
```
